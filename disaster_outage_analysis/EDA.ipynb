```python
earthquakes = pd.read_csv('earthquakes-2024-10-12_16-04-47_-0400.csv')
earthquakes
```

```python
print(earthquakes[['Year', 'Mo', 'Dy']].isnull().sum())
```

```python
earthquakes.dropna(subset=['Year', 'Mo', 'Dy'], inplace=True)
```

```python
print(earthquakes.columns)
print(earthquakes[['Year', 'Mo', 'Dy']].head())
print(earthquakes[['Year', 'Mo', 'Dy']].isnull().sum())
```

```python
earthquakes[['Year', 'Mo', 'Dy']].fillna(1, inplace=True)  # Fill NaN values with defaults
earthquakes[['Hr', 'Mn', 'Sec']].fillna(0, inplace=True)
```

```python
earthquakes['Year'] = earthquakes['Year'].astype(int)
earthquakes['Mo'] = earthquakes['Mo'].astype(int)
earthquakes['Dy'] = earthquakes['Dy'].astype(int)
```

```python
earthquakes['timestamp'] = pd.to_datetime(earthquakes['Year'].astype(str) + '-' + 
                                          earthquakes['Mo'].astype(str) + '-' + 
                                          earthquakes['Dy'].astype(str))
```

```python
earthquakes
```

```python
df['timestamp'] = pd.to_datetime(df['SEEN_NS'], errors='coerce')
```

```python
merged_data = pd.merge_asof(
    df.sort_values('timestamp'),
    earthquakes.sort_values('timestamp'),
    on='timestamp',
    direction='nearest',
    tolerance=pd.Timedelta(days=1)
)
```

```python
outages_near_disasters = merged_data[merged_data['IS_ONLINE'] == 0]
print(outages_near_disasters['Location Name'].value_counts())
```

```python
import matplotlib.pyplot as plt
outages_near_disasters.groupby('timestamp').size().plot(kind='bar', figsize=(12, 6))

# Setting the title
plt.title('Outages Near Earthquakes')

# Formatting the x-axis ticks (if the timestamps are datetime, you can format them to a simpler form)
plt.xticks(rotation=45, ha='right')

# Optionally, you can limit the number of x-ticks to avoid overlap
plt.tight_layout()  # Adjust layout to prevent label clipping

# Displaying the plot
plt.show()
```

```python
merged_data['Earthquake_Timestamp'] = pd.to_datetime(earthquakes['Year'].astype(str) + '-' + 
                                          earthquakes['Mo'].astype(str) + '-' + 
                                          earthquakes['Dy'].astype(str))
```

```python
print(merged_data[['timestamp', 'Earthquake_Timestamp']].isnull().sum())
```

```python
merged_data['time_diff'] = (
    merged_data['timestamp'] - merged_data['Earthquake_Timestamp']
).dt.total_seconds()
merged_data['time_diff'].hist(bins=50)
```

```python
print(merged_data[['time_diff', 'Mag']].corr())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

X = merged_data[['Mag', 'Focal Depth (km)', 'Latitude', 'Longitude']]
y = merged_data['IS_ONLINE']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = RandomForestClassifier()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Online', 'Offline'], yticklabels=['Online', 'Offline'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```

```python
features = X.columns
importances = model.feature_importances_
indices = np.argsort(importances)

plt.figure(figsize=(8, 6))
plt.title('Feature Importances')
plt.barh(range(len(indices)), importances[indices], align='center')
plt.yticks(range(len(indices)), [features[i] for i in indices])
plt.xlabel('Relative Importance')
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

# Compute ROC curve
fpr, tpr, thresholds = roc_curve(y_test, model.predict_proba(X_test)[:, 1])
roc_auc = auc(fpr, tpr)

# Plot ROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc='lower right')
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

# Hyperparameter tuning
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
}

grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best parameters
print("Best Parameters:", grid_search.best_params_)
```

```python
from geopy.distance import geodesic
merged_data_clean = merged_data.dropna(subset=['Latitude', 'Longitude'])
earthquakes_clean = earthquakes.dropna(subset=['Latitude', 'Longitude'])
distances = [
    geodesic((row['Latitude'], row['Longitude']), (eq['Latitude'], eq['Longitude'])).km
    for _, row in merged_data_clean.iterrows()
    for _, eq in earthquakes_clean.iterrows()
]

# Printing the distances
print(distances)
```

```python
jitter = np.random.normal(0, 0.2, size=len(distances))  # Small random noise
x_with_jitter = np.arange(len(distances)) + jitter
jitter = np.random.normal(0, 0.5, size=len(distances))  # Small random noise
x_with_jitter = np.arange(len(distances)) + jitter

# Plotting with reduced dot size
plt.figure(figsize=(10, 6))
plt.scatter(x_with_jitter, distances, color='blue', alpha=0.6, s=2)  # s=10 controls dot size
plt.title('Distances Between Locations and Earthquakes')
plt.xlabel('Index')
plt.ylabel('Distance (km)')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns

# Density plot of distances
plt.figure(figsize=(10, 6))
sns.kdeplot(distances, shade=True, color="skyblue")
plt.title('Density Plot of Distances Between Locations and Earthquakes')
plt.xlabel('Distance (km)')
plt.ylabel('Density')
plt.grid(True)
plt.show()
```

```python
import folium

# Create a base map
map = folium.Map(location=[merged_data_clean['Latitude'].mean(), merged_data_clean['Longitude'].mean()], zoom_start=5)

# Add earthquake markers
for _, eq in earthquakes_clean.iterrows():
    folium.CircleMarker(
        location=[eq['Latitude'], eq['Longitude']],
        radius=5,
        color='red',
        fill=True,
        fill_color='red'
    ).add_to(map)

# Add location markers
for _, row in merged_data_clean.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=5,
        color='blue',
        fill=True,
        fill_color='blue'
    ).add_to(map)

# Display the map
map
```
